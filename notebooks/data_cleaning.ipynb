## Reproducible final cleaning and review outputs

Run the following cell from any working directory. It reloads the original `DATA_MESSY` sheet, documents each deterministic correction, and writes (1) an analysis CSV containing valid positive-quantity records and (2) a review CSV containing all nonpositive-quantity rows with their original values. Exact duplicate rows are counted and deduplicated once; no invalid-quantity record is silently removed. Dollar signs are formatting and are safely removed from price strings. Revenue is retained as recorded. Revenue mismatches are quarantined for review because the 100 affected rows have invalid quantities and the original quantities cannot be established from this dataset alone. The other 149,016 unique, positive-quantity rows satisfy `Revenue = transaction_qty * unit_price` within one cent.

```python
from pathlib import Path
import pandas as pd

# Resolve the project root whether this notebook is launched from the root or notebooks/.
_here = Path.cwd()
PROJECT_ROOT = _here if (_here / "data" / "Coffee_Shop_Sales.xlsx").exists() else _here.parent
SOURCE_PATH = PROJECT_ROOT / "data" / "Coffee_Shop_Sales.xlsx"
CLEAN_PATH = PROJECT_ROOT / "data" / "coffee_shop_cleaned.csv"
REVIEW_PATH = PROJECT_ROOT / "data" / "coffee_shop_review.csv"

source_df = pd.read_excel(SOURCE_PATH, sheet_name="DATA_MESSY")
source_rows = len(source_df)
exact_duplicates = int(source_df.duplicated().sum())
work = source_df.drop_duplicates().copy()

# Normalize whitespace/case in labels. These transformations do not alter IDs.
for column in ["store_location", "product_category", "product_type", "product_detail"]:
    work[column] = work[column].astype("string").str.strip()
for column in ["store_location", "product_category"]:
    work[column] = work[column].str.title()

# Parse currency formatting explicitly; keep a count of values that remain invalid.
price_text = work["unit_price"].astype("string").str.strip()
price_numeric = pd.to_numeric(price_text.str.replace(r"[$,]", "", regex=True), errors="coerce")
price_failures = int((price_numeric.isna() & work["unit_price"].notna()).sum())
work["unit_price"] = price_numeric

# Mixed date strings are parsed flexibly; ISO date values remain unambiguous.
work["transaction_date"] = pd.to_datetime(
    work["transaction_date"].astype("string").str.strip(),
    format="mixed", dayfirst=True, errors="coerce"
)
work["transaction_time"] = pd.to_datetime(
    work["transaction_time"].astype("string").str.strip(),
    format="%H:%M:%S", errors="coerce"
).dt.time

# Use IDs to fill missing labels only after verifying every observed mapping is unique.
for id_column, label_column in [("product_id", "product_category"), ("store_id", "store_location")]:
    mapping_counts = work.dropna(subset=[label_column]).groupby(id_column)[label_column].nunique()
    ambiguous = mapping_counts[mapping_counts > 1]
    if not ambiguous.empty:
        raise ValueError(f"Ambiguous {id_column} to {label_column} mappings: {ambiguous.index.tolist()}")
    mapping = work.dropna(subset=[label_column]).drop_duplicates(id_column).set_index(id_column)[label_column]
    work[label_column] = work[label_column].fillna(work[id_column].map(mapping))

work["transaction_qty"] = pd.to_numeric(work["transaction_qty"], errors="coerce")
work["Revenue"] = pd.to_numeric(work["Revenue"], errors="coerce")

# Preserve every nonpositive quantity row, uncorrected, for investigation.
invalid_quantity = work["transaction_qty"].isna() | work["transaction_qty"].le(0)
review = work.loc[invalid_quantity].copy()
review["review_reason"] = "Quantity missing or nonpositive; quantity/revenue not inferred"
review["expected_revenue_if_quantity_valid"] = review["transaction_qty"] * review["unit_price"]
review["revenue_difference_if_quantity_valid"] = (
    review["Revenue"] - review["expected_revenue_if_quantity_valid"]
)
review.to_csv(REVIEW_PATH, index=False)

# Positive quantities with complete calculation inputs must reconcile; fail loudly otherwise.
calculable = work["transaction_qty"].gt(0) & work["unit_price"].notna() & work["Revenue"].notna()
work["expected_revenue"] = work["transaction_qty"] * work["unit_price"]
work["revenue_difference"] = work["Revenue"] - work["expected_revenue"]
mismatch = calculable & work["revenue_difference"].abs().gt(0.01)
if mismatch.any():
    raise ValueError(f"{int(mismatch.sum())} positive-quantity rows have revenue discrepancies; inspect before export")

# Keep only analytically valid records in the dashboard CSV; all quantity exceptions
# are retained, with original values, in REVIEW_PATH above.
clean = work.loc[~invalid_quantity].copy()
clean.to_csv(CLEAN_PATH, index=False)

print(f"Source rows: {source_rows:,}")
print(f"Exact duplicate rows removed: {exact_duplicates:,}")
print(f"Unique rows after exact deduplication: {len(work):,}")
print(f"Unit prices still failing conversion: {price_failures:,}")
print(f"Nonpositive/missing quantity review rows: {len(review):,}")
print(f"Positive-quantity revenue mismatches (> $0.01): {int(mismatch.sum()):,}")
print(f"Analysis CSV rows: {len(clean):,} -> {CLEAN_PATH}")
print(f"Review CSV rows: {len(review):,} -> {REVIEW_PATH}")
```
